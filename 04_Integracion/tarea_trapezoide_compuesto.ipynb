```python
import numpy as np

# PARÁMETROS GENERALES Y VALOR EXACTO

a = 8.0
b = 30.0


x_exacto = 11057 #obtenido previamente archivo tarea_regla_trapezoide.ipynb



# REGLA DEL TRAPECIO COMPUESTA PARA n = 2

n_2 = 2
h_2 = (b - a) / n_2

# Puntos de evaluación t0, t1, t2
t0 = a
t1 = a + 1.0 * h_2
t2 = b

# Evaluación de f(t) en cada punto
f_t0 = 2000.0 * np.log(140000.0 / (140000.0 - 2100.0 * t0)) - 9.8 * t0
f_t1 = 2000.0 * np.log(140000.0 / (140000.0 - 2100.0 * t1)) - 9.8 * t1
f_t2 = 2000.0 * np.log(140000.0 / (140000.0 - 2100.0 * t2)) - 9.8 * t2

# integral para n = 2
x_n2 = (h_2 / 2.0) * (f_t0 + 2.0 * f_t1 + f_t2)

# Errores
error_abs_n2 = np.abs(x_n2 - x_exacto)
error_rel_n2 = (error_abs_n2 / x_exacto) * 100.0

print("RESULTADOS PARA N = 2 SEGMENTOS")
print("a) x estimado:")
print(x_n2)
print("b) Error absoluto:")
print(error_abs_n2)
print("c) Error relativo:")
print(error_rel_n2)

```
