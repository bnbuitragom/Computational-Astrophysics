```python
import numpy as np

# PARÁMETROS DE INTEGRACIÓN
# Límite inferior a = 8, límite superior b = 30, n = 4 segmentos
a = 8.0
b = 30.0
n = 4

x_exacto = 11057 #obtenido previamente archivo tarea_regla_trapezoide.ipynb

h = (b - a) / n


# PUNTOS DE EVALUACIÓN t DE t0 A t4
t0 = a
t1 = a + 1.0 * h
t2 = a + 2.0 * h
t3 = a + 3.0 * h
t4 = b


# EVALUACIÓN DE LA FUNCIÓN INTEGRANDA EN CADA PUNTO
f_t0 = 2000.0 * np.log(140000.0 / (140000.0 - 2100.0 * t0)) - 9.8 * t0
f_t1 = 2000.0 * np.log(140000.0 / (140000.0 - 2100.0 * t1)) - 9.8 * t1
f_t2 = 2000.0 * np.log(140000.0 / (140000.0 - 2100.0 * t2)) - 9.8 * t2
f_t3 = 2000.0 * np.log(140000.0 / (140000.0 - 2100.0 * t3)) - 9.8 * t3
f_t4 = 2000.0 * np.log(140000.0 / (140000.0 - 2100.0 * t4)) - 9.8 * t4


# a) REGLA DE SIMPSON 1/3 (n = 4)
# Integral = (h / 3) * [ f(t0) + 4*f(t1) + 2*f(t2) + 4*f(t3) + f(t4) ]
x_simpson = (h / 3.0) * (f_t0 + 4.0 * f_t1 + 2.0 * f_t2 + 4.0 * f_t3 + f_t4)


# b) ERROR ABSOLUTO
error_absoluto = np.abs(x_simpson - x_exacto)


# c) ERROR RELATIVO
error_relativo = (error_absoluto / x_exacto) * 100.0


# RESULTADOS

print("VALOR EXACTO DE LA INTEGRAL")
print(x_exacto)

print("\na) ESTIMACIÓN POR REGLA DE SIMPSON 1/3 (N = 4)")
print(x_simpson)

print("\nb) ERROR ABSOLUTO")
print(error_absoluto)

print("\nc) ERROR RELATIVO")
print(error_relativo)
```
