```python
import numpy as np


# PARÁMETROS FÍSICOS
# L(t) = L0 * exp(-t / tau)

L0 = 5.0e42
tau = 111.4
t0 = 64.0
h = 2.0


# EVALUACIÓN DE PUNTOS PARA DIFERENCIAS FINITAS

L_minus3 = L0 * np.exp(-(t0 - 3.0 * h) / tau)
L_minus2 = L0 * np.exp(-(t0 - 2.0 * h) / tau)
L_minus1 = L0 * np.exp(-(t0 - h) / tau)
L_0      = L0 * np.exp(-t0 / tau)
L_plus1  = L0 * np.exp(-(t0 + h) / tau)
L_plus2  = L0 * np.exp(-(t0 + 2.0 * h) / tau)
L_plus3  = L0 * np.exp(-(t0 + 3.0 * h) / tau)



# SEGUNDA DERIVADA EXACTA Y EN DIFERENCIAS FINITAS L''(64)
# L''(t) = (L0 / tau^2) * exp(-t / tau)


exacta_L2 = (L0 / (tau**2)) * np.exp(-t0 / tau)

# Centrada O(h^2)
centered_L2 = (L_plus1 - 2.0 * L_0 + L_minus1) / (h**2)

# Hacia Adelante O(h^2)
forward_L2 = (-L_plus3 + 4.0 * L_plus2 - 5.0 * L_plus1 + 2.0 * L_0) / (h**2)

# Hacia Atrás O(h^2)
backward_L2 = (2.0 * L_0 - 5.0 * L_minus1 + 4.0 * L_minus2 - L_minus3) / (h**2)



# CÁLCULO DE ERRORES RELATIVOS PORCENTUALES


error_centered_L2 = np.abs((centered_L2 - exacta_L2) / exacta_L2) * 100.0
error_forward_L2 = np.abs((forward_L2 - exacta_L2) / exacta_L2) * 100.0
error_backward_L2 = np.abs((backward_L2 - exacta_L2) / exacta_L2) * 100.0



# RESULTADOS


print("SEGUNDA DERIVADA EXACTA L''(64)")
print(exacta_L2)

print("\nDIFERENCIAS FINITAS CENTRADA L''(64)")
print(centered_L2)

print("\nDIFERENCIAS FINITAS ADELANTE L''(64)")
print(forward_L2)

print("\nDIFERENCIAS FINITAS ATRÁS L''(64)")
print(backward_L2)

print("\nERROR RELATIVO CENTRADA")
print(error_centered_L2)

print("\nERROR RELATIVO ADELANTE")
print(error_forward_L2)

print("\nERROR RELATIVO ATRÁS")
print(error_backward_L2)
```
