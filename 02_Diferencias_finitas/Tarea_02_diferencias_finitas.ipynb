```python
import numpy as np


# PARÁMETROS FÍSICOS Y PUNTOS DE TIEMPO


h = 1.0

t_10 = 10.0
t_11 = 11.0
t_12 = 12.0
t_13 = 13.0
t_14 = 14.0



# 1. EVALUACIÓN DE F(t) EN t = 10, 11, 12, 13, 14
# F(t) = 100 + 15*sin(pi*t / 6) + 3*sin(pi*t / 2)


F_10 = 100.0 + 15.0 * np.sin(np.pi * t_10 / 6.0) + 3.0 * np.sin(np.pi * t_10 / 2.0)
F_11 = 100.0 + 15.0 * np.sin(np.pi * t_11 / 6.0) + 3.0 * np.sin(np.pi * t_11 / 2.0)
F_12 = 100.0 + 15.0 * np.sin(np.pi * t_12 / 6.0) + 3.0 * np.sin(np.pi * t_12 / 2.0)
F_13 = 100.0 + 15.0 * np.sin(np.pi * t_13 / 6.0) + 3.0 * np.sin(np.pi * t_13 / 2.0)
F_14 = 100.0 + 15.0 * np.sin(np.pi * t_14 / 6.0) + 3.0 * np.sin(np.pi * t_14 / 2.0)



# 2. CÁLCULO DE F'(12) USANDO DIFERENCIAS FINITAS (h = 1)


# Diferencia hacia adelante: ( F(13) - F(12) ) / h
F_prime_forward = (F_13 - F_12) / h

# Diferencia hacia atrás: ( F(12) - F(11) ) / h
F_prime_backward = (F_12 - F_11) / h

# Diferencia centrada: ( F(13) - F(11) ) / (2 * h)
F_prime_centered = (F_13 - F_11) / (2.0 * h)



# 3. DERIVADA ANALÍTICA F'(t) EN t = 12 Y ERRORES RELATIVOS
# F'(t) = 15*(pi/6)*cos(pi*t / 6) + 3*(pi/2)*cos(pi*t / 2)


F_prime_exact = 15.0 * (np.pi / 6.0) * np.cos(np.pi * t_12 / 6.0) + 3.0 * (np.pi / 2.0) * np.cos(np.pi * t_12 / 2.0)

# Error relativo porcentual para cada esquema
error_forward = np.abs((F_prime_forward - F_prime_exact) / F_prime_exact) * 100.0
error_backward = np.abs((F_prime_backward - F_prime_exact) / F_prime_exact) * 100.0
error_centered = np.abs((F_prime_centered - F_prime_exact) / F_prime_exact) * 100.0


# RESULTADOS


print("PUNTO 1: F(t) EN CADA TIEMPO")
print(F_10)
print(F_11)
print(F_12)
print(F_13)
print(F_14)

print("\nPUNTO 2: F'(12) APROXIMADA")
print(F_prime_forward)
print(F_prime_backward)
print(F_prime_centered)

print("\nPUNTO 3: DERIVADA EXACTA EN t = 12")
print(F_prime_exact)

print("\nERRORES RELATIVOS PORCENTUALES")
print(error_forward)
print(error_backward)
print(error_centered)
```
