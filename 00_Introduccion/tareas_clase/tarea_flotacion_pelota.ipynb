```python
import numpy as np


# PARÁMETROS FÍSICOS
R = 0.055
gamma_b = 0.6


# FUNCIÓN DE BÚSQUEDA DE RAÍCES: MÉTODO DE BISECCIÓN
def metodo_biseccion(a, b):
    tol = 1e-6 #ajustable
    max_iter = 100 #ajustable
    
    for i in range(max_iter):
        c = (a + b) / 2.0
        
        # Evaluacion de la funcion f(x) = x^3 - 3*R*x^2 + 4*(R^3)*gamma_b
        f_a = (a**3) - 3.0 * R * (a**2) + 4.0 * (R**3) * gamma_b
        f_c = (c**3) - 3.0 * R * (c**2) + 4.0 * (R**3) * gamma_b
        
        # Verificacion de convergencia
        if np.abs(f_c) < tol:
            return c
            
        # Actualizacion del intervalo
        if f_a * f_c < 0.0:
            b = c
        else:
            a = c
            
    return (a + b) / 2.0



# EJECUCIÓN Y RESULTADOS

a_inicial = 0.0
b_inicial = 2.0 * R

x_raiz = metodo_biseccion(a_inicial, b_inicial)

print("Las raices son:")
print(x_raiz)
```
