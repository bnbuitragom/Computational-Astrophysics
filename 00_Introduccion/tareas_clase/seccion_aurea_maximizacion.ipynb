```python
import numpy as np

xl = 0.0
xu = np.pi / 2.0
epsilon = 0.05

# Solucion teorica
theta_exacto = 1.0472 #rad

# Razon aurea (R)
R = (np.sqrt(5.0) - 1.0) / 2.0


# PASO 1: Puntos intermedios iniciales

d = R * (xu - xl)
x1 = xl + d
x2 = xu - d

#SECCIÓN ÁUREA PARA MAXIMIZACIÓN
iteracion = 0
error_actual = abs(xu - xl)

while error_actual >= epsilon:
    iteracion = iteracion + 1
    
    # PASO 2: Evaluar la funcion A(theta) = 4 * sin(theta) * (1 + cos(theta))
    f_x1 = 4.0 * np.sin(x1) * (1.0 + np.cos(x1))
    f_x2 = 4.0 * np.sin(x2) * (1.0 + np.cos(x2))
    
    # PASO 3: Actualizar el intervalo para Maximización
    if f_x1 > f_x2:
        xl = x2
        x2 = x1
        d = R * (xu - xl)
        x1 = xl + d
    else:
        xu = x1
        x1 = x2
        d = R * (xu - xl)
        x2 = xu - d
        
    # Criterio de tolerancia (ancho del intervalo actual)
    error_actual = abs(xu - xl)


# RESULTADO

theta_opt = (xl + xu) / 2.0
area_max = 4.0 * np.sin(theta_opt) * (1.0 + np.cos(theta_opt))

error_absoluto = abs(theta_opt - theta_exacto)
error_relativo_porcentual = (error_absoluto / theta_exacto) * 100.0



print("RESULTADOS DEL ALGORITMO DE SECCIÓN ÁUREA (MAXIMIZACIÓN)")
print("Numero de iteraciones realizadas:", iteracion)
print("Angulo optimo encontrado (theta*):", theta_opt, "rad")
print("Area maxima obtenida A(theta*):", area_max)
print("Intervalo final [xl, xu]: [", xl, ",", xu, "]")
print("Ancho del intervalo final (Tolerancia alcanzada):", error_actual)
print("Error absoluto respecto al valor teorico:", error_absoluto, "rad")
print("Error relativo:", error_relativo_porcentual, "%")
```
