```python
import numpy as np

# Puntos seleccionados alrededor de t = 21: (15, 362.78), (20, 517.35), (22.5, 602.97)

t_eval = 21.0

x0 = 15.0
f_x0 = 362.78

x1 = 20.0
f_x1 = 517.35

x2 = 22.5
f_x2 = 602.97


# CÁLCULO DE COEFICIENTES 

b0 = f_x0
b1 = (f_x1 - f_x0) / (x1 - x0)

diff_10 = (f_x1 - f_x0) / (x1 - x0)
diff_21 = (f_x2 - f_x1) / (x2 - x1)

b2 = (diff_21 - diff_10) / (x2 - x0)

# EVALUACIÓN DEL POLINOMIO f2(t) EN t = 21
# f2(x) = b0 + b1*(x - x0) + b2*(x - x0)*(x - x1)

term0 = b0
term1 = b1 * (t_eval - x0)
term2 = b2 * (t_eval - x0) * (t_eval - x1)
v_21 = term0 + term1 + term2

# RESULTADOS

print("COEFICIENTES B0, B1, B2")
print(b0)
print(b1)
print(b2)

print("\nVELOCIDAD EN t = 21")
print(v_21)
```
