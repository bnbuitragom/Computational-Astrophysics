```python
import numpy as np

# a) FUNCIÓN BALSICA DE BISECCIÓN PARA LA LEY DE WIEN
# f(x) = x - 5*(1 - exp(-x))


def bisection(xl, xu):
    # b) Criterio de Scarborough para n = 3 cifras significativas:
    # es = 0.5 * 10^(2 - 3) % = 0.05%
    es = 0.05
    
    xr = xl
    ea = 100.0
    
    while ea > es:
        xr_old = xr
        xr = (xl + xu) / 2.0
        
        # Evaluacion de f(x) en el limite inferior y en el punto medio
        f_xl = xl - 5.0 * (1.0 - np.exp(-xl))
        f_xr = xr - 5.0 * (1.0 - np.exp(-xr))
        
        # Calculo del error relativo porcentual aproximado
        if xr != 0.0:
            ea = np.abs((xr - xr_old) / xr) * 100.0
            
        # Actualizacion de limites del intervalo
        if f_xl * f_xr < 0.0:
            xu = xr
        elif f_xl * f_xr > 0.0:
            xl = xr
        else:
            break
            
    return xr



# PARÁMETROS INICIALES Y EJECUCIÓN
# Intervalo x en [0.5, 5]

xl_inicial = 0.5
xu_inicial = 5.0

raiz = bisection(xl_inicial, xu_inicial)

print("Las raices son:")
print(raiz)
```
