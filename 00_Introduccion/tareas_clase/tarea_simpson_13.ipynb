```python
import numpy as np

# Límites
a = 0.1
b = 1.3

# Integral  - (5/4) * exp(-2x) * (2x + 1)
F_b = -1.25 * np.exp(-2.0 * b) * (2.0 * b + 1.0)
F_a = -1.25 * np.exp(-2.0 * a) * (2.0 * a + 1.0)
x_exacto = F_b - F_a

# a) REGLA DE SIMPSON 1/3 
h = (b - a) / 2.0
x_pm = (a + b) / 2.0
f_a = 5.0 * a * np.exp(-2.0 * a)
f_pm = 5.0 * x_pm * np.exp(-2.0 * x_pm)
f_b = 5.0 * b * np.exp(-2.0 * b)
x_simpson = (h / 3.0) * (f_a + 4.0 * f_pm + f_b)

# b) ERROR ABSOLUTO
error_absoluto = np.abs(x_simpson - x_exacto)

# c) ERROR RELATIVO
error_relativo_porcentual = (error_absoluto / x_exacto) * 100.0

# RESULTADOS

print("VALOR EXACTO DE LA INTEGRAL")
print(x_exacto)

print("\na) ESTIMACIÓN POR SIMPSON 1/3 SIMPLE")
print(x_simpson)

print("\nb) ERROR ABSOLUTO")
print(error_absoluto)

print("\nc) ERROR RELATIVO PORCENTUAL")
print(error_relativo_porcentual)
```
