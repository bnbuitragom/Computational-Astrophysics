```python
import numpy as np


# PARÁMETROS FÍSICOS 

M_in = 5.0
tau_dep = 1.5
R = 0.4
h = 2.0

t_actual = 0.0
M_actual = 0.0

tol_porcentaje = 5.0
error_relativo = 100.0

# MÉTODO DE EULER 
while error_relativo > tol_porcentaje:
    # Evaluacion de la derivada dM/dt en el punto actual
    dM_dt = M_in - ((1.0 - R) * M_actual) / tau_dep
    
    # Paso de integracion de Euler: M_(i+1) = M_i + f(t_i, M_i) * h
    M_siguiente = M_actual + dM_dt * h
    
    # Calculo del error relativo porcentual
    error_relativo = np.abs((M_siguiente - M_actual) / M_siguiente) * 100.0
    
    # Actualizacion de variables
    t_actual = t_actual + h
    M_actual = M_siguiente

# RESULTADOS

print("Tiempo transcurrido:")
print(t_actual)

print("\nMasa de gas acumulada:")
print(M_actual)

print("\nError relativo alcanzado:")
print(error_relativo)
```

## Interpretación de Resultados

A partir de la simulación numérica utilizando el **Método de Euler**, se obtuvieron los siguientes valores al cumplir con el criterio de tolerancia ($\varepsilon_{\text{rel}} < 5\%$):

* **Tiempo transcurrido ($t$):** `6.0` Gy (Gigaaños)
* **Masa de gas acumulada ($M$):** `12.4` $M_\odot$ (Masas solares)
* **Error relativo alcanzado ($\varepsilon_{\text{rel}}$):** `3.23%`

### Análisis Físico:

1. **Acumulación de Gas:** La galaxia joven, partiendo de un estado inicial sin gas ($M(0) = 0$), logra acumular una masa de gas de **12.4 masas solares** al cabo de **6.0 Gigaaños** de evolución.
2. **Equilibrio Dinámico:** La tasa de acumulación disminuye gradualmente a medida que el gas acumulado se consume en la formación de estrellas y retroalimentación.
3. **Convergencia del Método:** El algoritmo se detiene en $t = 6.0$ Gy porque la variación relativa entre iteraciones consecutivas cae al **3.23%**, cumpliendo con la tolerancia establecida menor al 5%.
