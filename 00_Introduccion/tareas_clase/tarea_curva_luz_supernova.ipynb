```python
import numpy as np

# DATOS
# t: tiempo en dias
# L: flujo relativo (L/L0)


t = np.array([0.0, 10.0, 20.0, 30.0, 40.0, 50.0])
L = np.array([1.0, 0.914, 0.836, 0.764, 0.699, 0.639])

n = len(t)

# Logarirtmo
# z = ln(L)
z = np.log(L)

# CÁLCULO DE SUMATORIAS Y PROMEDIOS PARA REGRESIÓN LINEAL

sum_x = np.sum(t)
sum_z = np.sum(z)
sum_xz = np.sum(t * z)
sum_x2 = np.sum(t**2)

mean_x = sum_x / n
mean_z = sum_z / n

# a) CONSTANTES DE REGRESIÓN A Y lambda
# a1 = (n*sum_xz - sum_x*sum_z) / (n*sum_x2 - (sum_x)^2)
# a0 = mean_z - a1 * mean_x

a1 = (n * sum_xz - sum_x * sum_z) / (n * sum_x2 - (sum_x**2))
a0 = mean_z - a1 * mean_x

A = np.exp(a0)
lambda_const = -a1


# b) VIDA MEDIA DEL 56Co
# t_mitad = ln(2) / lambda

t_mitad = np.log(2.0) / lambda_const



# c) FLUJO RELATIVO DESPUÉS DE 100 DÍAS
# L_100 = A * exp(-lambda * 100)

t_futuro = 100.0
L_100 = A * np.exp(-lambda_const * t_futuro)



# RESULTADOS

print("a) CONSTANTE A")
print(A)

print("\na) CONSTANTE LAMBDA")
print(lambda_const)

print("\nb) VIDA MEDIA")
print(t_mitad)

print("\nc) FLUJO RELATIVO A LOS 100 DIAS")
print(L_100)
```
