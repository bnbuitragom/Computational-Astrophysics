```python
import numpy as np

# 1. beta Aquarii (Sadalsuud)
h1 = 21.0
m1 = 31.0
s1 = 33.53
d1 = -5.0
dm1 = 34.0
ds1 = 16.2

# 2. eta Piscium (Alpherg)
h2 = 1.0
m2 = 31.0
s2 = 29.01
d2 = 15.0
dm2 = 20.0
ds2 = 45.0
```

```python
# Estrella 1
m1_en_horas = m1 / 60.0
s1_en_horas = s1 / 3600.0
suma_hms_1 = h1 + m1_en_horas + s1_en_horas
ra1_deg = 15.0 * suma_hms_1
alpha_1 = np.deg2rad(ra1_deg)

# Estrella 2
m2_en_horas = m2 / 60.0
s2_en_horas = s2 / 3600.0
suma_hms_2 = h2 + m2_en_horas + s2_en_horas
ra2_deg = 15.0 * suma_hms_2
alpha_2 = np.deg2rad(ra2_deg)

print("ra1_deg:", ra1_deg, "| alpha_1 (rad):", alpha_1)
print("ra2_deg:", ra2_deg, "| alpha_2 (rad):", alpha_2)
```

```python
# Estrella 1
dm1_en_grados = dm1 / 60.0
ds1_en_grados = ds1 / 3600.0
menos_d1 = -1.0 * d1
suma_dms_1 = menos_d1 + dm1_en_grados + ds1_en_grados
dec1_deg = -1.0 * suma_dms_1
delta_1 = np.deg2rad(dec1_deg)

# Estrella 2
dm2_en_grados = dm2 / 60.0
ds2_en_grados = ds2 / 3600.0
dec2_deg = d2 + dm2_en_grados + ds2_en_grados
delta_2 = np.deg2rad(dec2_deg)

print("dec1_deg:", dec1_deg, "| delta_1 (rad):", delta_1)
print("dec2_deg:", dec2_deg, "| delta_2 (rad):", delta_2)
```

```python
# Diferencias angulares
diff_delta = delta_1 - delta_2
abs_diff_delta = np.abs(diff_delta)
medio_diff_delta = abs_diff_delta / 2.0

diff_alpha = alpha_1 - alpha_2
abs_diff_alpha = np.abs(diff_alpha)
medio_diff_alpha = abs_diff_alpha / 2.0

# Termino 1: sin^2(|delta_1 - delta_2| / 2)
seno_delta = np.sin(medio_diff_delta)
haversine_delta = seno_delta ** 2

# Termino 2: cos(delta_1) * cos(delta_2) * sin^2(|alpha_1 - alpha_2| / 2)
cos_d1 = np.cos(delta_1)
cos_d2 = np.cos(delta_2)
seno_alpha = np.sin(medio_diff_alpha)
haversine_alpha = seno_alpha ** 2
producto_cos_sen = cos_d1 * cos_d2 * haversine_alpha

# Raiz y distancia angular final
suma_haversine = haversine_delta + producto_cos_sen
raiz_haversine = np.sqrt(suma_haversine)
arcoseno_val = np.arcsin(raiz_haversine)
distancia_rad = 2.0 * arcoseno_val
distancia_deg = np.rad2deg(distancia_rad)

print("Distancia angular (rad):", distancia_rad)
print("Distancia angular (deg):", distancia_deg)
```
