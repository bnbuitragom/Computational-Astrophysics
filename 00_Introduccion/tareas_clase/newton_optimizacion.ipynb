```python
import numpy as np

x_i = np.radians(45.0)
epsilon = 0.05
theta_exacto = 1.0472

iteracion = 0
error_relativo = 1.0
f_prime_val = 1.0

#Algoritmo de Newton
while abs(f_prime_val) >= epsilon and error_relativo >= epsilon:
    iteracion = iteracion + 1
    
    # Calculo de derivadas en el punto actual x_i
    f_prime_val = 4.0 * np.cos(x_i) + 4.0 * np.cos(2.0 * x_i)
    f_dprime_val = -4.0 * np.sin(x_i) - 8.0 * np.sin(2.0 * x_i)
    
    # Iteracion: x_{i+1} = x_i - f'(x_i) / f''(x_i)
    x_i_next = x_i - (f_prime_val / f_dprime_val)
    
    # Criterio de tolerancia: error relativo entre pasos
    error_relativo = abs((x_i_next - x_i) / x_i_next)
    
    # Actualizacion de la estimacion
    x_i = x_i_next

# Evaluacion de f(x_{i+1}) en la solucion encontrada
theta_opt = x_i
area_max = 4.0 * np.sin(theta_opt) * (1.0 + np.cos(theta_opt))

error_absoluto = abs(theta_opt - theta_exacto)
error_relativo = (error_absoluto / theta_exacto) * 100.0

#Resultados
print("RESULTADOS DEL ALGORITMO DE NEWTON PARA OPTIMIZACIÓN")
print("Se encontro el valor de theta en la iteracion:", iteracion)
print("Angulo optimo encontrado (theta*):", theta_opt, "rad")
print("Angulo optimo en grados:", np.degrees(theta_opt), "deg")
print("Area maxima obtenida A(theta*):", area_max)
print("Valor de f'(theta*) alcanzado:", f_prime_val)
print("Error relativo porcentual respecto al valor exacto:", error_relativo, "%")
```
