```python
import numpy as np

# PARÁMETROS DE INTEGRACIÓN 
a = 8.0
b = 30.0

x_exacto = 11061.34

# EVALUACIÓN DE LA FUNCIÓN
# f(t) = 2000 * ln(140000 / (140000 - 2100*t)) - 9.8*t
f_a = 2000.0 * np.log(140000.0 / (140000.0 - 2100.0 * a)) - 9.8 * a
f_b = 2000.0 * np.log(140000.0 / (140000.0 - 2100.0 * b)) - 9.8 * b

# a) INTEGRACIÓN POR REGLA DEL TRAPEZOIDE
x_trapezoide = (b - a) * (f_a + f_b) / 2.0

# b) ERROR ABSOLUTO
error_absoluto = np.abs(x_trapezoide - x_exacto)

# c) ERROR RELATIVO 
error_relativo_porcentual = np.abs((x_trapezoide - x_exacto) / x_exacto) * 100.0

# RESULTADOS


print("INTEGRAL ESTIMADA POR TRAPEZOIDE")
print(x_trapezoide)

print("\nERROR ABSOLUTO")
print(error_absoluto)

print("\nERROR RELATIVO PORCENTUAL")
print(error_relativo_porcentual)
```
