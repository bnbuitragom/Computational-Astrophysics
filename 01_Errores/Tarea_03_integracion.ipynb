```python
import numpy as np

# Coeficientes del polinomio alpha(T) = a0 + a1*T + a2*T^2
a0 = 1.0827e-5
a1 = 1.11503e-8
a2 = -2.21004e-11

# Temperatura ambiente
T_room = 26.0


# a) Hielo seco (T_fluid = -77)

T_fluid_a = -77.0

# I(T) = a0*T + (a1/2)*T^2 + (a2/3)*T^3
I_room_a = a0 * T_room + (a1 / 2.0) * (T_room**2) + (a2 / 3.0) * (T_room**3)
I_fluid_a = a0 * T_fluid_a + (a1 / 2.0) * (T_fluid_a**2) + (a2 / 3.0) * (T_fluid_a**3)

# Integral definida (evaluada por T_fluid - T_room para delta D / D)
integral_a = I_fluid_a - I_room_a


# b) Nitrógeno líquido (T_fluid = -196)

T_fluid_b = -196.0

# Evaluación en T_fluid_b
I_room_b = a0 * T_room + (a1 / 2.0) * (T_room**2) + (a2 / 3.0) * (T_room**3)
I_fluid_b = a0 * T_fluid_b + (a1 / 2.0) * (T_fluid_b**2) + (a2 / 3.0) * (T_fluid_b**3)

# Integral definida
integral_b = I_fluid_b - I_room_b

# -------------------------------------------------------------
# RESULTADOS
# -------------------------------------------------------------
print("RESULTADOS")
print(f"Integral a (Hielo seco -77C): {integral_a:.6e}")
print(f"Integral b (Nitrogeno liquido -196C): {integral_b:.6e}")
```
