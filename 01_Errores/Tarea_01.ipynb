# Tarea: Expansión en series de Taylor

Datos que nos da el problema:
A = 0.1, ω = 3.105, t = 1, Δt = 0.2

La curva de luz es:
y(t) = A sin(ωt)

Y su expansión en series de Taylor es:

A sin(tω) = A t ω · Σ_{k=0}^∞ [(-1)^k (ωt)^(2k)] / (1+2k)!

```python
import math

# Datos que nos da el ejercicio
A = 0.1
omega = 3.105
t = 1

x = omega * t
```

## a) Suma truncada en k=6 y en k=7


Cada término tiene la forma:

término_k = A * t * omega * [ (-1)^k * (omega*t)^(2k) ] / (1+2k)!

```python
suma6 = 0


for k in range(0, 7):
    numerador = ((-1)**k) * (x**(2*k))
    denominador = math.factorial(1 + 2*k)
    termino = A * t * omega * (numerador / denominador)
    
    suma6 = suma6 + termino

print("Suma truncada en k=6:", suma6)
```

```python
# Hacemos lo mismo pero ahora hasta k=7
suma7 = 0

for k in range(0, 8):
    numerador = ((-1)**k) * (x**(2*k))
    denominador = math.factorial(1 + 2*k)
    termino = A * t * omega * (numerador / denominador)
    
    suma7 = suma7 + termino

print("Suma truncada en k=7:", suma7)
```

## b) Error aproximado

Usamos la fórmula:

ε_approx = (suma7 - suma6) / suma7

Lo multiplicamos por 100 para tenerlo en porcentaje.

```python
error_aprox = (suma7 - suma6) / suma7
error_aprox_porcentaje = abs(error_aprox) * 100

print("Error aproximado:", error_aprox)
print("Error aproximado en porcentaje:", error_aprox_porcentaje)
```

## c) Criterio de Scarborough

ε_approx <= 0.5 × 10^(2-m)


```python
# Despejamos m usando la fórmula que obtuvimos
m = 2 - math.log10(error_aprox_porcentaje / 0.5)

print("Valor de m (sin redondear):", m)

# Como el número de cifras significativas debe ser un entero,
# tomamos la parte entera hacia abajo (floor)
m_final = math.floor(m)

print("Número de cifras significativas garantizadas:", m_final)
```

## d) Para la casa

Ahora queremos encontrar para qué valor de k la serie ya tiene
7 cifras significativas correctas, usando el mismo criterio de
Scarborough pero con m=7.

La idea es: vamos aumentando k de uno en uno, calculamos la suma
truncada en ese k y en el k anterior, sacamos el error aproximado,
y nos detenemos cuando el error ya sea suficientemente pequeño.

```python
m_deseado = 7
limite_deseado = 0.5 * 10**(2 - m_deseado)

print("Límite que necesitamos alcanzar:", limite_deseado)

k = 1

while True:
    # suma truncada hasta k
    suma_actual = 0
    for i in range(0, k + 1):
        numerador = ((-1)**i) * (x**(2*i))
        denominador = math.factorial(1 + 2*i)
        termino = A * t * omega * (numerador / denominador)
        suma_actual = suma_actual + termino
    
    # suma truncada hasta k-1
    suma_anterior = 0
    for i in range(0, k):
        numerador = ((-1)**i) * (x**(2*i))
        denominador = math.factorial(1 + 2*i)
        termino = A * t * omega * (numerador / denominador)
        suma_anterior = suma_anterior + termino
    
    # calculamos el error aproximado en este k
    error_actual = abs((suma_actual - suma_anterior) / suma_actual) * 100
    
    print("k =", k, "-> suma =", suma_actual, "-> error =", error_actual)
    
    if error_actual <= limite_deseado:
        break   # ya llegamos a la precisión que queríamos
    
    k = k + 1

print("\nValor de k necesario para 7 cifras significativas correctas:", k)
print("Suma en ese k:", suma_actual)
```
