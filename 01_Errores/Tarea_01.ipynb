# Tarea: Curva de luz

Datos que nos da el problema:
A = 0.1, ω = 3.105, t = 1

La curva de luz es:
y(t) = A sin(ωt)

Y su expansión en series de Taylor es:

$$y(t) = A \sin(\omega t) = A t \omega \sum_{k=0}^{\infty} \frac{(-1)^k (\omega t)^{2k}}{(1 + 2k)!}$$

```python
import math

# Datos que nos da el ejercicio
A = 0.1
omega = 3.105
t = 1

x = omega * t
```

## a) Suma truncada en k=6 y en k=7


Cada término tiene la forma:


$$
\text{término}_k = A \cdot \omega \cdot t \cdot \frac{(-1)^k \cdot (\omega \cdot t)^{2k}}{(2k+1)!}
$$

```python
suma6 = 0


for k in range(0, 7):
    numerador = ((-1)**k) * (x**(2*k))
    denominador = math.factorial(1 + 2*k)
    termino = A * t * omega * (numerador / denominador)
    
    suma6 = suma6 + termino

print("Suma truncada en k=6:", suma6)
```

```python
# Hacemos lo mismo pero ahora hasta k=7
suma7 = 0

for k in range(0, 8):
    numerador = ((-1)**k) * (x**(2*k))
    denominador = math.factorial(1 + 2*k)
    termino = A * t * omega * (numerador / denominador)
    
    suma7 = suma7 + termino

print("Suma truncada en k=7:", suma7)
```

## b) Error aproximado

Usamos la fórmula:

ε_approx = (suma7 - suma6) / suma7

Lo multiplicamos por 100 para tenerlo en porcentaje.

```python
error_aprox = (suma7 - suma6) / suma7
error_aprox_porcentaje = abs(error_aprox) * 100

print("Error aproximado:", error_aprox)
print("Error aproximado en porcentaje:", error_aprox_porcentaje)
```

## c) Criterio de Scarborough

ε_approx <= 0.5 × 10^(2-m)


```python
# Despejamos m 
m = 2 - math.log10(error_aprox_porcentaje / 0.5)

print("Valor de m (sin redondear):", m)

# Como el número de cifras significativas debe ser un entero,

m_final = math.floor(m) # Función que da el mayor entero menor o igual al número

print("Número de cifras significativas garantizadas:", m_final)
```

## d) Para la casa

Ahora usamos lA misma fórmula, pero en vez de calcular m una sola para calcular para cada valor de k.

```python
# Empezamos con la suma en 0 (todavía sin sumar ningún término)
suma_anterior = 0
k = 0

while True:
    # calculamos el término que le toca a este k
    numerador = ((-1)**k) * (x**(2*k))
    denominador = math.factorial(1 + 2*k)
    termino = A * t * omega * (numerador / denominador)
    
    # lo sumamos a lo que ya teníamos, y esa es la suma actual
    suma_actual = suma_anterior + termino
    
    # si k=0 todavía no podemos calcular el error (no hay suma "anterior" real)
    if k > 0:
        error_actual = abs((suma_actual - suma_anterior) / suma_actual) * 100
        m_actual = 2 - math.log10(error_actual / 0.5)
        m_actual_entero = math.floor(m_actual)
        
        print("k =", k, "-> error =", error_actual, "-> m =", m_actual_entero)
        
        if m_actual_entero >= 7:
            break
    
    # preparamos la siguiente vuelta: lo que era "actual" ahora es "anterior"
    suma_anterior = suma_actual
    k = k + 1

print("\nValor de k necesario para 7 cifras significativas correctas:", k)
print("Suma en ese k:", suma_actual)
```
