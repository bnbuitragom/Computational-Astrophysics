```python
import numpy as np

# Parametros del problema
epsilon = 0.05
R = (np.sqrt(5.0) - 1.0) / 2.0
l_l = 0.0
l_u = 9.0
theta_0 = np.pi / 6.0

# Puntos intermedios para l
d_l = R * (l_u - l_l)
l1 = l_l + d_l
l2 = l_u - d_l

# Evaluacion inicial de la funcion en l1 y l2
f_l1 = (18.0 - 2.0 * l1 + 2.0 * l1 * np.cos(theta_0)) * l1 * np.sin(theta_0)
f_l2 = (18.0 - 2.0 * l2 + 2.0 * l2 * np.cos(theta_0)) * l2 * np.sin(theta_0)

iter_l = 0

while (l_u - l_l) >= epsilon:
    iter_l = iter_l + 1
    
    if f_l1 > f_l2:
        l_l = l2
        l2 = l1
        f_l2 = f_l1
        d_l = R * (l_u - l_l)
        l1 = l_l + d_l
        f_l1 = (18.0 - 2.0 * l1 + 2.0 * l1 * np.cos(theta_0)) * l1 * np.sin(theta_0)
    else:
        l_u = l1
        l1 = l2
        f_l1 = f_l2
        d_l = R * (l_u - l_l)
        l2 = l_u - d_l
        f_l2 = (18.0 - 2.0 * l2 + 2.0 * l2 * np.cos(theta_0)) * l2 * np.sin(theta_0)

# l optimo encontrado
l_opt = (l_l + l_u) / 2.0


#Iterar sobre theta con l_opt fijo
theta_l = 0.0
theta_u = np.pi / 2.0

# Puntos intermedios para theta
d_theta = R * (theta_u - theta_l)
theta1 = theta_l + d_theta
theta2 = theta_u - d_theta

# Evaluacion inicial de la funcion en theta1 y theta2
f_th1 = (18.0 - 2.0 * l_opt + 2.0 * l_opt * np.cos(theta1)) * l_opt * np.sin(theta1)
f_th2 = (18.0 - 2.0 * l_opt + 2.0 * l_opt * np.cos(theta2)) * l_opt * np.sin(theta2)

iter_theta = 0

while (theta_u - theta_l) >= epsilon:
    iter_theta = iter_theta + 1
    
    if f_th1 > f_th2:
        theta_l = theta2
        theta2 = theta1
        f_th2 = f_th1
        d_theta = R * (theta_u - theta_l)
        theta1 = theta_l + d_theta
        f_th1 = (18.0 - 2.0 * l_opt + 2.0 * l_opt * np.cos(theta1)) * l_opt * np.sin(theta1)
    else:
        theta_u = theta1
        theta1 = theta2
        f_th1 = f_th2
        d_theta = R * (theta_u - theta_l)
        theta2 = theta_u - d_theta
        f_th2 = (18.0 - 2.0 * l_opt + 2.0 * l_opt * np.cos(theta2)) * l_opt * np.sin(theta2)

# theta optimo y area maxima final
theta_opt = (theta_l + theta_u) / 2.0
area_max = (18.0 - 2.0 * l_opt + 2.0 * l_opt * np.cos(theta_opt)) * l_opt * np.sin(theta_opt)
b_opt = 18.0 - 2.0 * l_opt


# RESULTADOS
print("RESULTADOS DE LA OPTIMIZACION")
print("Iteraciones necesarias en coordenada l:", iter_l)
print("Longitud de lado optima l_opt:", l_opt, "cm")
print("Iteraciones necesarias en coordenada theta:", iter_theta)
print("Angulo optimo theta_opt:", theta_opt, "rad")
print("Angulo optimo en grados:", np.degrees(theta_opt), "deg")
print("Base optima del canal b_opt:", b_opt, "cm")
print("Area maxima conseguida f(l_opt, theta_opt):", area_max, "cm^2")
```
