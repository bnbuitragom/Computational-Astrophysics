```python
import numpy as np

# PARÁMETROS FÍSICOS

M_max = 1
tau = 1
h = 0.01

t_inicio = 0
t_final = 5 * tau

pasos = int((t_final - t_inicio) / h)

t = t_inicio
M = 0.0

# INTEGRACIÓN CON RK4

i = 0

while i < pasos:
    # k1
    k1 = (M_max - M) / tau
    
    # k2
    M_k2 = M + 0.5 * h * k1
    k2 = (M_max - M_k2) / tau
    
    #k3
    M_k3 = M + 0.5 * h * k2
    k3 = (M_max - M_k3) / tau
    
    # k4
    M_k4 = M + h * k3
    k4 = (M_max - M_k4) / tau
    
    # M_(i+1)
    M = M + (h / 6.0) * (k1 + 2.0 * k2 + 2.0 * k3 + k4)
    
    # Avanzar tiempo y contador
    t = t + h
    i = i + 1


# RESULTADO


print(M)
```
