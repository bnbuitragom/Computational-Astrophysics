```python
import numpy as np

# PARÁMETROS FÍSICOS E INICIALES
e = 0.7
M = 2.0
E0 = M


# DEFINICIÓN DE LA FUNCIÓN Y SU DERIVADA
# f(E) = E - e*sin(E) - M
# f_prima(E) = 1 - e*cos(E)

def f(E):
    return E - e * np.sin(E) - M

def f_prima(E):
    return 1.0 - e * np.cos(E)



# CRITERIO DE SCARBOROUGH
m_cifras = 3
es = 0.5 * (10.0 ** (2.0 - m_cifras))


# FUNCIÓN NEWTON-RAPHSON
def NewtonRaphson(x0, f, f_prima):
    xr = x0
    ea = 100.0
    
    while ea > es:
        xr_old = xr
        
        # Newton-Raphson: x_(n+1) = x_n - f(x_n) / f'(x_n)
        xr = xr_old - (f(xr_old) / f_prima(xr_old))
        
        #error relativo
        if xr != 0.0:
            ea = np.abs((xr - xr_old) / xr) * 100.0
            
    return xr


# RESULTADOS
E_raiz = NewtonRaphson(E0, f, f_prima)

print("las raices son")
print(E_raiz)
```
